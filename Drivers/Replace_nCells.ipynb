```python
import numpy as np
import xarray as xr
```

```python
f='/glade/campaign/cgd/amp/juliob/mpasa3p75km/TimeInvariant/zgrid_dyamond_nCells.nc'
X=xr.open_dataset(f)
```

```python
X
```

```python
coords = dict( 
    ncol = ( ["ncol"],X.ncol.values ),
)

Xo = xr.Dataset( coords=coords  )



lon=X.lon.values
lat=X.lat.values
zgrid=X.zgrid.values.T
zgrid=np.flip( zgrid , axis=0 )


Dar = xr.DataArray( data=lat , 
                    dims=('ncol',),
                    attrs=dict( long_name='latitude',) ,) 


Xo['lat']=Dar

Dar = xr.DataArray( data=lon , 
                    dims=('ncol',),
                    attrs=dict( long_name='longitude',) ,) 


Xo['lon']=Dar

Dar = xr.DataArray( data=zgrid , 
                    dims=('lev','ncol',),
                    attrs=dict( long_name='altitude of grid edges (m)',) ,) 


Xo['zgrid']=Dar

```

```python
Xo
```

```python

fo='/glade/campaign/cgd/amp/juliob/mpasa3p75km/TimeInvariant/zgrid_dyamond_ncol.nc'
Xo.to_netcdf( fo )
```

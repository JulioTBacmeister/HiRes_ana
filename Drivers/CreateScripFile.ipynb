```python
def nice_domain_name(latr=None,lonr=None):

    if latr is None or lonr is None:
        sys.exit("ERROR: no valid input provided")
        
    ltS,ltN = latr
    lnW,lnE = lonr

    if ltN < 0:
        ltN_ = f'{abs(ltN)}S'
    elif ltN > 0:
        ltN_ = f'{abs(ltN)}N'
    elif ltN == 0:
        ltN_ = f'{abs(ltN)}'
    if ltS < 0:
        ltS_ = f'{abs(ltS)}S'
    elif ltS > 0:
        ltS_ = f'{abs(ltS)}N'
    elif ltS == 0:
        ltS_ = f'{abs(ltS)}'

    lnE_ = f'{lnE}'
    lnW_ = f'{lnW}'

    return f"Lat_{ltS_}-{ltN_}_Lon_{lnW_}-{lnE_}"
    
        
```
